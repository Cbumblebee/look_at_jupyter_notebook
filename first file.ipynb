## Useful shortcuts

- shift + enter takes you to the next cell after running it
- ctrl + enter runs the cell but stays there
- esc takes u out of a cell
- enter will enter the cell
- after escaping out of a cell, you can add a cell above it by typing A
- below the cell with B
- delete a cell with DD
- moving up and down: J/K
- change between markdown/code types of cells: esc out, then type M for markdown, Y for code
